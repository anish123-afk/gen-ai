```python
from langchain_community.tools import tool
```

```python
def multiply(a, b):
    """Multiple two numbers"""
    return a * b
```

```python
def multiply(a : int , b : int) -> int:
    """Multiple two numbers"""

    return a * b
```

```python
@tool
def multiply(a : int , b : int) -> int:
    """Multiple two numbers"""

    return a * b
```

```python
result = multiply.invoke({
    'a' : 3,
    'b' : 5
})
```
